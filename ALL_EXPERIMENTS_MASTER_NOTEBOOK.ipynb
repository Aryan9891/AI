# Artificial Intelligence Lab - LAB 01
## Experiment 1: Vacuum Cleaner World Agent (2x1 Grid)
### Objective:
Design a Vacuum Cleaner World Agent (2x1 Grid) using Python (Function-based, No Classes).
- Environment: 2x1 Grid with dirt placed randomly (0 = Clean, 1 = Dirty).
- Knowledge State: Separate from the hidden environment state. Initialized as [None, None] (?).
- Perception: Agent only senses cell status when physically visiting it.
- Actions: Move Left/Right (only if current cell is clean), Vacuum (if dirty).
- Stopping condition: Terminate when agent's knowledge confirms all cells are clean, or step limit reached.

### 1. Environment and Knowledge Representation
Functions:
- `create_environment()`: returns initial 2x1 array `[0, 0]`
- `place_dirt(env)`: randomly sets dirt status (1 or 0)
- `place_agent(grid_size)`: randomly selects starting index
- `create_knowledge(grid_size)`: returns belief state `[None, None]`

```python
import random

def create_environment():
    """Creates and returns a 2x1 array representing true hidden state."""
    return [0, 0]

def place_dirt(env):
    """Randomly places dirt (1) or clean (0) in cells."""
    for i in range(len(env)):
        env[i] = random.choice([0, 1])
    if sum(env) == 0:
        env[random.randint(0, len(env) - 1)] = 1
    return env

def place_agent(grid_size=2):
    """Randomly selects and returns a starting position."""
    return random.randint(0, grid_size - 1)

def create_knowledge(grid_size=2):
    """Creates agent's internal knowledge array initialized to None."""
    return [None] * grid_size
```

### 2. Sensing and Action Functions
- `perceive(env, knowledge, pos)`: Senses true status of occupied cell.
- `is_dirty(env, pos)`: Returns True if cell is dirty.
- `vacuum(env, knowledge, pos)`: Cleans current cell in env and updates knowledge.
- `get_possible_moves(pos, grid_size)`: Boundary-checked valid directions.
- `move_agent(pos, grid_size)`: Moves agent to an adjacent valid cell.

```python
def perceive(env, knowledge, pos):
    knowledge[pos] = env[pos]
    return knowledge

def is_dirty(env, pos):
    return env[pos] == 1

def vacuum(env, knowledge, pos):
    env[pos] = 0
    knowledge[pos] = 0
    return env, knowledge

def get_possible_moves(pos, grid_size=2):
    moves = []
    if pos > 0:
        moves.append('Left')
    if pos < grid_size - 1:
        moves.append('Right')
    return moves

def move_agent(pos, grid_size=2):
    valid_moves = get_possible_moves(pos, grid_size)
    if not valid_moves:
        return pos, None
    direction = random.choice(valid_moves)
    new_pos = pos - 1 if direction == 'Left' else pos + 1
    return new_pos, direction
```

### 3. Termination Checks and Display Formatter
- `knowledge_all_clean(knowledge)`: Checks if agent's belief indicates all cells are clean.
- `all_clean(env)`: Ground-truth environment check.
- `display_environment(...)`: Prints state matching lab notation.

```python
def knowledge_all_clean(knowledge):
    for status in knowledge:
        if status is None or status == 1:
            return False
    return True

def all_clean(env):
    return all(cell == 0 for cell in env)

def display_environment(knowledge, agent_pos, action, step):
    header = 'Initial Stage:' if step == 0 else f'Run {step}:'
    cells_str = ''
    for i in range(len(knowledge)):
        status = knowledge[i]
        status_label = '?' if status is None else ('D' if status == 1 else 'C')
        if i == agent_pos:
            if action == 'perceive' or action is None:
                cell_repr = f'[A(),{status_label}]'
            elif action == 'vacuum':
                cell_repr = f'[A(V),{status_label}]'
            elif action == 'move_Right':
                cell_repr = f'[A(R),{status_label}]'
            elif action == 'move_Left':
                cell_repr = f'[A(L),{status_label}]'
            else:
                cell_repr = f'[A({action}),{status_label}]'
        else:
            cell_repr = '[?]' if status is None else f'[{status_label}]'
        cells_str += cell_repr
    print(header)
    print(cells_str)
```

### 4. Running Experiment 1 (2x1 Grid)

```python
def run_vacuum_world(max_iterations=20):
    env = create_environment()
    env = place_dirt(env)
    agent_pos = place_agent(len(env))
    knowledge = create_knowledge(len(env))

    perceive(env, knowledge, agent_pos)
    step = 0
    display_environment(knowledge, agent_pos, 'perceive', step)

    while not knowledge_all_clean(knowledge) and step < max_iterations:
        step += 1
        if knowledge[agent_pos] == 1:
            vacuum(env, knowledge, agent_pos)
            display_environment(knowledge, agent_pos, 'vacuum', step)
        else:
            new_pos, direction = move_agent(agent_pos, len(env))
            action_label = f'move_{direction}'
            display_environment(knowledge, agent_pos, action_label, step)
            agent_pos = new_pos
            perceive(env, knowledge, agent_pos)
            step += 1
            display_environment(knowledge, agent_pos, 'perceive', step)

    if knowledge_all_clean(knowledge):
        print('All cells are clean! Task Completed.')
    else:
        print('Maximum iterations reached.')
    print(f'True environment state: {env} (All clean: {all_clean(env)})')

# Seed for reproducible demonstration
random.seed(42)
run_vacuum_world()
```

### 5. Experiment 2: Generalized N-Cell Grid Simulation

```python
def run_vacuum_world_generalized(grid_size=4, max_iterations=50):
    print(f'=== Running Generalized Vacuum World (Size {grid_size}x1) ===')
    env = [0] * grid_size
    env = place_dirt(env)
    agent_pos = place_agent(grid_size)
    knowledge = create_knowledge(grid_size)

    perceive(env, knowledge, agent_pos)
    step = 0
    display_environment(knowledge, agent_pos, 'perceive', step)

    while not knowledge_all_clean(knowledge) and step < max_iterations:
        step += 1
        if knowledge[agent_pos] == 1:
            vacuum(env, knowledge, agent_pos)
            display_environment(knowledge, agent_pos, 'vacuum', step)
        else:
            new_pos, direction = move_agent(agent_pos, grid_size)
            action_label = f'move_{direction}'
            display_environment(knowledge, agent_pos, action_label, step)
            agent_pos = new_pos
            perceive(env, knowledge, agent_pos)
            step += 1
            display_environment(knowledge, agent_pos, 'perceive', step)

    if knowledge_all_clean(knowledge):
        print('All cells are clean! Task Completed.')
    else:
        print('Maximum iterations reached.')
    print(f'True environment state: {env}')

random.seed(42)
run_vacuum_world_generalized(grid_size=4)
```

# Artificial Intelligence Lab - LAB 02
## Experiment 2: Utility-Based Vacuum Cleaner Agent (m × n Grid)
### Objective:
Design a Utility-Based Vacuum Cleaner Agent operating on an $m \times n$ grid with performance metric evaluation:
1. **Fully Observable Environment**: The agent can see every cell's status, computes exact BFS distances to remaining dirt, and selects actions maximizing utility.
2. **Partially Observable Environment**: The agent operates with incomplete knowledge, maintaining a belief state.

### Utility Function:
$$U(a) = \text{VacuumReward}(a) - \text{DistanceToNearestDirty}(a) - \text{MovementCost}(a) + \text{DirtDensityNearby}(a)$$

### 1. Environment and State Representation Functions
- `create_environment(rows, cols)`
- `place_dirt(env, dirt_prob)`
- `place_agent(rows, cols)`
- `is_dirty(env, pos)`
- `vacuum(env, pos)`
- `get_possible_moves(pos, rows, cols)`

```python
import random
from collections import deque

def create_environment(rows, cols):
    """Creates a 2D matrix (rows x cols) filled with 0 (Clean)."""
    return [[0 for _ in range(cols)] for _ in range(rows)]

def place_dirt(env, dirt_prob=0.4):
    """Randomly sets cells to 1 (Dirty) based on dirt_prob."""
    rows, cols = len(env), len(env[0])
    dirt_count = 0
    for r in range(rows):
        for c in range(cols):
            if random.random() < dirt_prob:
                env[r][c] = 1
                dirt_count += 1
    if dirt_count == 0:
        env[0][0] = 1
        env[rows - 1][cols - 1] = 1
    return env

def place_agent(rows, cols):
    """Randomly places agent at a starting (row, col) position."""
    return (random.randint(0, rows - 1), random.randint(0, cols - 1))

def is_dirty(env, pos):
    r, c = pos
    return env[r][c] == 1

def vacuum(env, pos):
    r, c = pos
    env[r][c] = 0
    return env

def get_possible_moves(pos, rows, cols):
    r, c = pos
    moves = {}
    if r > 0:
        moves['Up'] = (r - 1, c)
    if r < rows - 1:
        moves['Down'] = (r + 1, c)
    if c > 0:
        moves['Left'] = (r, c - 1)
    if c < cols - 1:
        moves['Right'] = (r, c + 1)
    return moves
```

### 2. Distance Computation (BFS) and Utility Evaluation
- `find_nearest_dirty_cell`: Computes shortest BFS distance.
- `get_dirt_density_nearby`: Counts dirty cells in local radius.
- `compute_utility`: Applies the exact utility formula.
- `choose_best_action`: Selects candidate action with argmax $U(a)$.

```python
def find_nearest_dirty_cell(env, pos, rows, cols):
    """Uses BFS shortest-path search to locate the nearest dirty cell and distance."""
    if is_dirty(env, pos):
        return pos, 0

    queue = deque([(pos[0], pos[1], 0)])
    visited = {pos}

    while queue:
        r, c, dist = queue.popleft()
        if env[r][c] == 1:
            return (r, c), dist

        for nr, nc in [(r - 1, c), (r + 1, c), (r, c - 1), (r, c + 1)]:
            if 0 <= nr < rows and 0 <= nc < cols and (nr, nc) not in visited:
                visited.add((nr, nc))
                queue.append((nr, nc, dist + 1))

    return None, 0

def get_dirt_density_nearby(env, pos, rows, cols, radius=1):
    r, c = pos
    density = 0
    for dr in range(-radius, radius + 1):
        for dc in range(-radius, radius + 1):
            nr, nc = r + dr, c + dc
            if 0 <= nr < rows and 0 <= nc < cols:
                if env[nr][nc] == 1:
                    density += 1
    return density

def compute_utility(action, pos, env, rows, cols):
    if action == 'vacuum':
        resulting_pos = pos
        vacuum_reward = 10 if is_dirty(env, pos) else -5
        movement_cost = 0
    else:
        moves = get_possible_moves(pos, rows, cols)
        resulting_pos = moves[action]
        vacuum_reward = 0
        movement_cost = 1

    _, dist_to_dirty = find_nearest_dirty_cell(env, resulting_pos, rows, cols)
    density = get_dirt_density_nearby(env, resulting_pos, rows, cols, radius=1)

    utility = vacuum_reward - dist_to_dirty - movement_cost + density
    return utility

def choose_best_action(pos, env, rows, cols):
    candidate_actions = []
    if is_dirty(env, pos):
        candidate_actions.append('vacuum')
    else:
        moves = get_possible_moves(pos, rows, cols)
        candidate_actions.extend(moves.keys())

    if not candidate_actions:
        return None

    best_action = None
    best_utility = float('-inf')

    for a in candidate_actions:
        u = compute_utility(a, pos, env, rows, cols)
        if u > best_utility:
            best_utility = u
            best_action = a

    return best_action, best_utility
```

### 3. Metric Tracking and Visualizers

```python
def all_clean(env):
    return all(cell == 0 for row in env for cell in row)

def compute_theoretical_minimum(env, start_pos, rows, cols):
    dirty_cells = [(r, c) for r in range(rows) for c in range(cols) if env[r][c] == 1]
    if not dirty_cells:
        return 0
    max_dist = max(abs(start_pos[0] - r) + abs(start_pos[1] - c) for r, c in dirty_cells)
    return len(dirty_cells) + max_dist

def update_metrics(metrics, action_taken):
    metrics['total_steps'] += 1
    if action_taken == 'vacuum':
        metrics['vacuums'] += 1
        metrics['energy_consumed'] += 2
    else:
        metrics['moves'] += 1
        metrics['energy_consumed'] += 1
    return metrics

def display_environment(env, agent_pos, action, step, rows, cols):
    header = 'Initial Stage:' if step == 0 else f'Step {step} (Action: {action}):'
    print(header)
    for r in range(rows):
        row_str = ''
        for c in range(cols):
            status = 'D' if env[r][c] == 1 else 'C'
            if (r, c) == agent_pos:
                if action == 'vacuum':
                    cell_repr = f'[A(V),{status}]'
                elif action in ['Up', 'Down', 'Left', 'Right']:
                    abbr = action[0]
                    cell_repr = f'[A({abbr}),{status}]'
                else:
                    cell_repr = f'[A(),{status}]'
            else:
                cell_repr = f'[{status}]'
            row_str += cell_repr
        print(row_str)
    print()

def display_performance_summary(metrics):
    print('=' * 45)
    print('        PERFORMANCE SUMMARY REPORT')
    print('=' * 45)
    print(f"Total Steps Executed:      {metrics['total_steps']}")
    print(f"Total Clean/Vacuum Actions:{metrics['vacuums']}")
    print(f"Total Movement Steps:      {metrics['moves']}")
    print(f"Total Energy Consumed:     {metrics['energy_consumed']} units")
    print(f"Theoretical Min Steps:     {metrics['theoretical_min']}")
    if metrics['theoretical_min'] > 0:
        ratio = metrics['total_steps'] / metrics['theoretical_min']
        print(f"Optimality Ratio:          {ratio:.2f} (Actual / Theoretical)")
    print('=' * 45)
```

### 4. Running Fully Observable Simulation ($3 \times 3$ Grid)

```python
def run_vacuum_world(rows=3, cols=3, max_iterations=50):
    env = create_environment(rows, cols)
    env = place_dirt(env, dirt_prob=0.35)
    agent_pos = place_agent(rows, cols)

    metrics = {
        'total_steps': 0,
        'vacuums': 0,
        'moves': 0,
        'energy_consumed': 0,
        'theoretical_min': compute_theoretical_minimum(env, agent_pos, rows, cols)
    }

    display_environment(env, agent_pos, 'None', 0, rows, cols)

    step = 0
    while not all_clean(env) and step < max_iterations:
        step += 1
        best_act, util = choose_best_action(agent_pos, env, rows, cols)

        if best_act == 'vacuum':
            vacuum(env, agent_pos)
            update_metrics(metrics, 'vacuum')
            display_environment(env, agent_pos, 'vacuum', step, rows, cols)
        else:
            moves = get_possible_moves(agent_pos, rows, cols)
            agent_pos = moves[best_act]
            update_metrics(metrics, best_act)
            display_environment(env, agent_pos, best_act, step, rows, cols)

    if all_clean(env):
        print('Success: All cells in grid are clean! Task Completed.')
    else:
        print('Stopped: Maximum iterations reached.')

    display_performance_summary(metrics)

random.seed(42)
run_vacuum_world(rows=3, cols=3)
```

### 5. Running Partially Observable Utility Agent (Belief State)

```python
def run_partially_observable_utility_agent(rows=3, cols=3, max_iterations=60):
    env = create_environment(rows, cols)
    env = place_dirt(env, dirt_prob=0.35)
    agent_pos = place_agent(rows, cols)
    knowledge = [[None for _ in range(cols)] for _ in range(rows)]

    r, c = agent_pos
    knowledge[r][c] = env[r][c]

    step = 0
    print(f'Initial State - Agent at {agent_pos}:')

    def belief_all_clean(k):
        return all(cell == 0 for row in k for cell in row if cell is not None) and                all(cell is not None for row in k for cell in row)

    while not belief_all_clean(knowledge) and step < max_iterations:
        step += 1
        r, c = agent_pos
        knowledge[r][c] = env[r][c]

        if knowledge[r][c] == 1:
            vacuum(env, agent_pos)
            knowledge[r][c] = 0
            action = 'vacuum'
        else:
            moves = get_possible_moves(agent_pos, rows, cols)
            best_move = None
            best_score = float('-inf')
            for act, npos in moves.items():
                nr, nc = npos
                score = 0
                if knowledge[nr][nc] is None:
                    score += 5
                elif knowledge[nr][nc] == 1:
                    score += 10
                else:
                    score -= 1
                if score > best_score:
                    best_score = score
                    best_move = act
            if best_move is None:
                best_move = random.choice(list(moves.keys()))
            agent_pos = moves[best_move]
            action = best_move

        print(f'Step {step}: Action = {action}, Agent Pos = {agent_pos}')

    print('Completed exploration & cleaning in Partially Observable World.')

random.seed(42)
run_partially_observable_utility_agent(rows=3, cols=3)
```

### 1. Data Structures & Mathematical Rationale
**Why 0 is wrong for 'no edge'**:
In weighted graphs, 0 represents a valid edge of zero cost (free transition). Using 0 for absent edges leads algorithms like Dijkstra to consider disconnected vertices as freely accessible, corrupting shortest paths. We must use `math.inf`.

```python
import math
import heapq
from collections import deque
import matplotlib.pyplot as plt
import networkx as nx

nodes = ['A', 'B', 'C', 'D', 'E', 'F']
node_to_idx = {n: i for i, n in enumerate(nodes)}

# 1. Adjacency List (Dictionary of neighbor, weight tuples)
adj_list = {
    'A': [('B', 4), ('C', 4)],
    'B': [('A', 4), ('C', 2), ('E', 1)],
    'C': [('A', 4), ('B', 2), ('D', 1)],
    'D': [('C', 1), ('F', 2)],
    'E': [('B', 1), ('F', 4)],
    'F': [('D', 2), ('E', 4)]
}

# 2. Adjacency Matrix (with math.inf sentinel)
adj_matrix = [
    #  A         B         C         D         E         F
    [0,        4,        4,        math.inf, math.inf, math.inf],  # A
    [4,        0,        2,        math.inf, 1,        math.inf],  # B
    [4,        2,        0,        1,        math.inf, math.inf],  # C
    [math.inf, math.inf, 1,        0,        math.inf, 2],         # D
    [math.inf, 1,        math.inf, math.inf, 0,        4],         # E
    [math.inf, math.inf, math.inf, 2,        4,        0]          # F
]

graph_pos = {
    'A': (0.0, 1.0),
    'B': (1.0, 1.6),
    'C': (1.0, 0.4),
    'D': (2.0, 2.0),
    'E': (2.0, 0.5),
    'F': (2.8, 1.3)
}
```

### 2. Plotting Graphs from Both Representations

```python
def plot_graph_from_adj_list():
    G = nx.Graph()
    for u, neighbors in adj_list.items():
        for v, w in neighbors:
            G.add_edge(u, v, weight=w)

    plt.figure(figsize=(7, 5))
    nx.draw_networkx_nodes(G, graph_pos, node_color='#6baed6', node_size=1200, edgecolors='black')
    nx.draw_networkx_labels(G, graph_pos, font_size=14, font_weight='bold', font_color='white')
    nx.draw_networkx_edges(G, graph_pos, width=2, edge_color='#555555')
    edge_labels = nx.get_edge_attributes(G, 'weight')
    nx.draw_networkx_edge_labels(G, graph_pos, edge_labels=edge_labels, font_size=12)
    plt.title("Visual Graph Representation (from Adjacency List)", fontsize=13, fontweight='bold')
    plt.axis('off')
    plt.show()

def plot_graph_from_adj_matrix():
    G = nx.Graph()
    n = len(nodes)
    for i in range(n):
        for j in range(i + 1, n):
            w = adj_matrix[i][j]
            if w != math.inf and w > 0:
                G.add_edge(nodes[i], nodes[j], weight=int(w))

    plt.figure(figsize=(7, 5))
    nx.draw_networkx_nodes(G, graph_pos, node_color='#74c476', node_size=1200, edgecolors='black')
    nx.draw_networkx_labels(G, graph_pos, font_size=14, font_weight='bold', font_color='white')
    nx.draw_networkx_edges(G, graph_pos, width=2, edge_color='#555555')
    edge_labels = nx.get_edge_attributes(G, 'weight')
    nx.draw_networkx_edge_labels(G, graph_pos, edge_labels=edge_labels, font_size=12)
    plt.title("Visual Graph Representation (from Adjacency Matrix)", fontsize=13, fontweight='bold')
    plt.axis('off')
    plt.show()

plot_graph_from_adj_list()
plot_graph_from_adj_matrix()
```

### 3. Implementation of BFS, DFS, and Dijkstra's Algorithms

```python
def get_path_cost(adj, path):
    cost = 0
    for u, v in zip(path[:-1], path[1:]):
        for nbr, w in adj[u]:
            if nbr == v:
                cost += w
                break
    return cost

def bfs_path(adj, start='A', goal='F'):
    queue = deque([[start]])
    visited = {start}
    visit_order = [start]

    while queue:
        path = queue.popleft()
        node = path[-1]
        if node == goal:
            return path, get_path_cost(adj, path), visit_order

        for nbr, _ in sorted(adj[node], key=lambda x: x[0]):
            if nbr not in visited:
                visited.add(nbr)
                visit_order.append(nbr)
                queue.append(path + [nbr])

    return None, 0, visit_order

def dfs_path(adj, start='A', goal='F'):
    stack = [[start]]
    visited = set()
    visit_order = []

    while stack:
        path = stack.pop()
        node = path[-1]
        if node not in visited:
            visited.add(node)
            visit_order.append(node)
            if node == goal:
                return path, get_path_cost(adj, path), visit_order

            for nbr, _ in sorted(adj[node], key=lambda x: x[0], reverse=True):
                if nbr not in visited:
                    stack.append(path + [nbr])

    return None, 0, visit_order

def dijkstra_path(adj, start='A', goal='F'):
    dist = {n: math.inf for n in adj}
    parent = {n: None for n in adj}
    dist[start] = 0
    pq = [(0, start)]
    visit_order = []
    visited = set()

    while pq:
        d, u = heapq.heappop(pq)
        if u in visited:
            continue
        visited.add(u)
        visit_order.append(u)
        if u == goal:
            break

        for v, weight in sorted(adj[u], key=lambda x: x[0]):
            if dist[u] + weight < dist[v]:
                dist[v] = dist[u] + weight
                parent[v] = u
                heapq.heappush(pq, (dist[v], v))

    path = []
    curr = goal
    while curr is not None:
        path.append(curr)
        curr = parent[curr]
    path.reverse()

    return path, dist[goal], visit_order
```

### 4. Running Searches and Comparing Results

```python
bfs_res = bfs_path(adj_list, 'A', 'F')
dfs_res = dfs_path(adj_list, 'A', 'F')
dijkstra_res = dijkstra_path(adj_list, 'A', 'F')

results = [
    ("Breadth-First Search (BFS)", bfs_res),
    ("Depth-First Search (DFS)", dfs_res),
    ("Dijkstra's Algorithm", dijkstra_res)
]

print("-" * 75)
print(f"{'Algorithm Name':<30} | {'Path':<20} | {'Edge Count':<10} | {'Total Cost':<10}")
print("-" * 75)
for name, (path, cost, order) in results:
    path_str = " -> ".join(path)
    edge_count = len(path) - 1
    print(f"{name:<30} | {path_str:<20} | {edge_count:<10} | {cost:<10}")
print("-" * 75)
```

## Experiment 4: Informed Searches (Greedy Best-First & A* Search)
### Objectives:

### 1. Data Structures and Heuristics Table

### 3. Implementation of Greedy Best-First Search and A* Search

```python
def get_path_cost(adj, path):
    cost = 0
    for u, v in zip(path[:-1], path[1:]):
        for nbr, w in adj[u]:
            if nbr == v:
                cost += w
                break
    return cost

def best_first_path(adj, h_dict, start='A', goal='F'):
    pq = [(h_dict[start], start, [start])]
    visited = set()
    visit_order = []

    while pq:
        h_val, u, path = heapq.heappop(pq)
        if u in visited:
            continue
        visited.add(u)
        visit_order.append(u)

        if u == goal:
            return path, get_path_cost(adj, path), visit_order

        for v, _ in sorted(adj[u], key=lambda x: x[0]):
            if v not in visited:
                heapq.heappush(pq, (h_dict[v], v, path + [v]))

    return None, 0, visit_order

def astar_path(adj, h_dict, start='A', goal='F'):
    pq = [(h_dict[start], 0, start, [start])]
    g_score = {node: math.inf for node in adj}
    g_score[start] = 0
    visited = set()
    visit_order = []

    while pq:
        f_val, g_val, u, path = heapq.heappop(pq)
        if u in visited:
            continue
        visited.add(u)
        visit_order.append(u)

        if u == goal:
            return path, g_val, visit_order

        for v, weight in sorted(adj[u], key=lambda x: x[0]):
            tentative_g = g_val + weight
            if tentative_g < g_score[v]:
                g_score[v] = tentative_g
                f_val_v = tentative_g + h_dict[v]
                heapq.heappush(pq, (f_val_v, tentative_g, v, path + [v]))

    return None, 0, visit_order
```

## Experiment 5: Local Search (Hill Climbing & Simulated Annealing)
### Objectives:
1. **Continuous 1D Function Global Optimization**:
   - Find global maximum of $f(x) = \sin(x) + \sin(10x/3)$ for $x \in [0, 10]$.
   - Compare **Hill Climbing** vs. **Simulated Annealing**.
   - Visualize function landscape and step traces.
2. **8-Queens Problem Formulation**:
   - Objective function: Non-attacking queen pairs ($28 - \text{conflicts}$).
   - Compare **First-Choice Hill Climbing** vs. **Simulated Annealing**.

### 1. Part 1: Continuous Function Optimization Functions

```python
import math
import random
import numpy as np
import matplotlib.pyplot as plt

def objective_function(x):
    return math.sin(x) + math.sin(10.0 * x / 3.0)

def hill_climbing_continuous(x_init=1.5, step_size=0.05, max_iter=200):
    current_x = x_init
    current_f = objective_function(current_x)
    history = [(current_x, current_f)]

    for _ in range(max_iter):
        cand_left = max(0.0, min(10.0, current_x - step_size))
        cand_right = max(0.0, min(10.0, current_x + step_size))

        f_left = objective_function(cand_left)
        f_right = objective_function(cand_right)

        if f_left > current_f and f_left >= f_right:
            current_x, current_f = cand_left, f_left
            history.append((current_x, current_f))
        elif f_right > current_f:
            current_x, current_f = cand_right, f_right
            history.append((current_x, current_f))
        else:
            break

    return current_x, current_f, history

def simulated_annealing_continuous(x_init=1.5, t_init=50.0, alpha=0.995, max_iter=2000):
    current_x = x_init
    current_f = objective_function(current_x)
    best_x, best_f = current_x, current_f
    history = [(current_x, current_f)]

    t = t_init
    for _ in range(max_iter):
        step = random.uniform(-4.0, 4.0)
        cand_x = max(0.0, min(10.0, current_x + step))
        cand_f = objective_function(cand_x)

        delta = cand_f - current_f

        if delta > 0 or math.exp(delta / max(t, 1e-8)) > random.random():
            current_x, current_f = cand_x, cand_f
            history.append((current_x, current_f))
            if current_f > best_f:
                best_x, best_f = current_x, current_f

        t *= alpha
        if t < 1e-4:
            break

    return best_x, best_f, history
```

### 2. Part 1: Optimization Results and Visual Comparison

```python
def visualize_continuous_optimization(hc_hist, sa_hist):
    x_vals = np.linspace(0, 10, 1000)
    y_vals = [objective_function(x) for x in x_vals]

    fig, axes = plt.subplots(1, 2, figsize=(15, 5))

    hc_xs, hc_ys = zip(*hc_hist)
    axes[0].plot(x_vals, y_vals, label='f(x)', color='#2b8cbe', linewidth=2)
    axes[0].plot(hc_xs, hc_ys, 'o-', color='#e34a33', markersize=4, alpha=0.7, label='HC Steps')
    axes[0].scatter([hc_xs[0]], [hc_ys[0]], color='green', s=100, zorder=5, label=f'Start ({hc_xs[0]:.2f})')
    axes[0].scatter([hc_xs[-1]], [hc_ys[-1]], color='red', s=120, marker='*', zorder=5, label=f'Trapped ({hc_xs[-1]:.2f})')
    axes[0].set_title(f"Hill Climbing: Trapped at Local Max ({hc_ys[-1]:.3f})", fontweight='bold')
    axes[0].set_xlabel("x")
    axes[0].set_ylabel("f(x)")
    axes[0].grid(True, linestyle='--', alpha=0.5)
    axes[0].legend()

    sa_xs, sa_ys = zip(*sa_hist)
    axes[1].plot(x_vals, y_vals, label='f(x)', color='#2b8cbe', linewidth=2)
    axes[1].plot(sa_xs, sa_ys, 'o-', color='#8856a7', markersize=3, alpha=0.3, label='SA Trajectory')
    axes[1].scatter([sa_xs[0]], [sa_ys[0]], color='green', s=100, zorder=5, label=f'Start ({sa_xs[0]:.2f})')
    best_idx = np.argmax(sa_ys)
    axes[1].scatter([sa_xs[best_idx]], [sa_ys[best_idx]], color='gold', s=150, marker='*', edgecolors='black', zorder=5, label=f'Global Max ({sa_xs[best_idx]:.2f})')
    axes[1].set_title(f"Simulated Annealing: Reached Global Max ({sa_ys[best_idx]:.3f})", fontweight='bold')
    axes[1].set_xlabel("x")
    axes[1].set_ylabel("f(x)")
    axes[1].grid(True, linestyle='--', alpha=0.5)
    axes[1].legend()

    plt.tight_layout()
    plt.show()

random.seed(1)
hc_x, hc_f, hc_hist = hill_climbing_continuous(x_init=1.5)
sa_x, sa_f, sa_hist = simulated_annealing_continuous(x_init=1.5)
print(f"Hill Climbing:       x = {hc_x:.4f}, f(x) = {hc_f:.4f}")
print(f"Simulated Annealing: x = {sa_x:.4f}, f(x) = {sa_f:.4f}")
visualize_continuous_optimization(hc_hist, sa_hist)
```

### 3. Part 2: 8-Queens Problem Formulation & Algorithms

```python
def get_8queens_score(state):
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return 28 - conflicts

def get_random_neighbor_8queens(state):
    neighbor = list(state)
    col = random.randint(0, 7)
    available_rows = [r for r in range(8) if r != state[col]]
    neighbor[col] = random.choice(available_rows)
    return neighbor

def first_choice_hill_climbing_8queens(max_iter=1000, max_tries_per_step=100):
    current_state = [random.randint(0, 7) for _ in range(8)]
    current_score = get_8queens_score(current_state)
    score_history = [current_score]

    for step in range(max_iter):
        if current_score == 28:
            break

        found_better = False
        for _ in range(max_tries_per_step):
            neighbor = get_random_neighbor_8queens(current_state)
            score = get_8queens_score(neighbor)
            if score > current_score:
                current_state = neighbor
                current_score = score
                score_history.append(current_score)
                found_better = True
                break

        if not found_better:
            break

    return current_state, current_score, score_history

def simulated_annealing_8queens(t_init=20.0, alpha=0.995, max_iter=5000):
    current_state = [random.randint(0, 7) for _ in range(8)]
    current_score = get_8queens_score(current_state)
    best_state = list(current_state)
    best_score = current_score
    score_history = [current_score]

    t = t_init
    for _ in range(max_iter):
        if best_score == 28:
            break

        neighbor = get_random_neighbor_8queens(current_state)
        cand_score = get_8queens_score(neighbor)
        delta = cand_score - current_score

        if delta > 0 or math.exp(delta / max(t, 1e-8)) > random.random():
            current_state = neighbor
            current_score = cand_score
            score_history.append(current_score)
            if current_score > best_score:
                best_state = list(current_state)
                best_score = current_score

        t *= alpha
        if t < 1e-4:
            break

    return best_state, best_score, score_history

def print_8queens_board(state):
    n = len(state)
    print("  +-" + "--" * n + "+")
    for r in range(n):
        row_str = f"{8 - r} | "
        for c in range(n):
            row_str += "Q " if state[c] == r else ". "
        row_str += "|"
        print(row_str)
    print("  +-" + "--" * n + "+")
    print("    A B C D E F G H")
    print(f"Non-attacking pairs: {get_8queens_score(state)} / 28 (Conflicts: {28 - get_8queens_score(state)})\n")
```

### 4. Part 2: Running 8-Queens Local Search Algorithms

```python
random.seed(101)
fc_state, fc_score, fc_hist = first_choice_hill_climbing_8queens()
sa_state, sa_score, sa_hist = simulated_annealing_8queens()

print("--- First-Choice Hill Climbing ---")
print_8queens_board(fc_state)

print("--- Simulated Annealing ---")
print_8queens_board(sa_state)
```

# Artificial Intelligence Lab - LAB 06
## Experiment 6: Adversarial Search (Tic-Tac-Toe with Minimax)
### Objectives:
1. Implement a two-player Tic-Tac-Toe game with a Line User Interface (LUI).
2. Modify the program so that one player is an AI Agent utilizing the **Minimax Algorithm** (game tree search).
3. Ensure input validation, game over announcement, and robust terminal display.

### 1. Board Representation and Helper Functions
- `create_board()`: Creates empty 9-cell board.
- `display_board()`: Renders board showing numbers 1-9 for empty cells.
- `check_winner()`: Checks 8 winning combinations or draw.
- `is_valid_move()`: Validates boundary and cell occupancy.

```python
def create_board():
    return [' ' for _ in range(9)]

def display_board(board):
    def cell(idx):
        return str(idx + 1) if board[idx] == ' ' else board[idx]

    print(f"{cell(0)} | {cell(1)} | {cell(2)}")
    print("---+---+---")
    print(f"{cell(3)} | {cell(4)} | {cell(5)}")
    print("---+---+---")
    print(f"{cell(6)} | {cell(7)} | {cell(8)}")

def check_winner(board):
    win_combinations = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)
    ]
    for a, b, c in win_combinations:
        if board[a] == board[b] == board[c] and board[a] != ' ':
            return board[a]
    if ' ' not in board:
        return 'Draw'
    return None

def is_valid_move(board, move):
    if move in range(1, 10):
        return board[move - 1] == ' '
    return False
```

### 2. Minimax Adversarial Search Algorithm
The AI computes the best move by evaluating the entire game tree. Faster wins are rewarded with higher scores using depth discounting.

```python
def minimax(board, depth, is_maximizing, ai_player='O', human_player='X'):
    state = check_winner(board)
    if state == ai_player:
        return 10 - depth, None
    elif state == human_player:
        return depth - 10, None
    elif state == 'Draw':
        return 0, None

    if is_maximizing:
        best_score = float('-inf')
        best_move = None
        for i in range(9):
            if board[i] == ' ':
                board[i] = ai_player
                score, _ = minimax(board, depth + 1, False, ai_player, human_player)
                board[i] = ' '
                if score > best_score:
                    best_score = score
                    best_move = i
        return best_score, best_move
    else:
        best_score = float('inf')
        best_move = None
        for i in range(9):
            if board[i] == ' ':
                board[i] = human_player
                score, _ = minimax(board, depth + 1, True, ai_player, human_player)
                board[i] = ' '
                if score < best_score:
                    best_score = score
                    best_move = i
        return best_score, best_move

def get_computer_move(board, ai_player='O', human_player='X'):
    _, best_move_idx = minimax(board, 0, True, ai_player, human_player)
    return best_move_idx + 1
```

### 3. Game Driver and Demo Simulation

```python
def play_against_ai(preset_inputs=None):
    print("--- TIC-TAC-TOE ---")
    input_iter = iter(preset_inputs) if preset_inputs else None

    def get_input(prompt):
        if input_iter:
            val = next(input_iter)
            print(f"{prompt}{val}")
            return val
        return input(prompt).strip()

    play_first = get_input("Do you want to play first? (y/n): ").lower()
    human = 'X' if play_first == 'y' else 'O'
    ai = 'O' if human == 'X' else 'X'

    print(f"You are {human}, the computer is {ai}.")
    board = create_board()
    display_board(board)

    current = 'X'
    while True:
        if current == human:
            while True:
                try:
                    move = int(get_input(f"Your move ({human}): "))
                    if is_valid_move(board, move):
                        board[move - 1] = human
                        break
                    print("Invalid or occupied square! Try again.")
                except ValueError:
                    print("Please enter a valid digit (1-9).")
        else:
            print("Computer is thinking...")
            comp_move = get_computer_move(board, ai_player=ai, human_player=human)
            print(f"Computer plays {comp_move}")
            board[comp_move - 1] = ai
            display_board(board)

        winner = check_winner(board)
        if winner:
            if current == human and winner != 'Draw':
                display_board(board)
            if winner == 'Draw':
                print("The game is a draw!")
            elif winner == ai:
                print(f"Computer ({ai}) wins!")
            else:
                print(f"Congratulations! You ({human}) win!")
            break

        current = 'O' if current == 'X' else 'X'

# Replay exact sample sequence from assignment sheet
sample_moves = ['y', '1', '9', '4']
play_against_ai(preset_inputs=sample_moves)
```

### 1. Board Representation, Gravity, and Win Detection

```python
ROWS = 4
COLS = 4

def create_board():
    return [['-' for _ in range(COLS)] for _ in range(ROWS)]

def display_board(board):
    print("| 1 | 2 | 3 | 4 |")
    print("|:-:|:-:|:-:|:-:|")
    for r in range(ROWS):
        print("| " + " | ".join(board[r]) + " |")

def get_lowest_empty_row(board, col):
    for r in range(ROWS - 1, -1, -1):
        if board[r][col] == '-':
            return r
    return -1

def is_valid_column(board, col):
    if 0 <= col < COLS:
        return board[0][col] == '-'
    return False

def check_win(board, player):
    # Horizontal
    for r in range(ROWS):
        for c in range(COLS - 2):
            if board[r][c] == board[r][c + 1] == board[r][c + 2] == player:
                return True
    # Vertical
    for c in range(COLS):
        for r in range(ROWS - 2):
            if board[r][c] == board[r + 1][c] == board[r + 2][c] == player:
                return True
    # Diagonal \
    for r in range(ROWS - 2):
        for c in range(COLS - 2):
            if board[r][c] == board[r + 1][c + 1] == board[r + 2][c + 2] == player:
                return True
    # Diagonal /
    for r in range(2, ROWS):
        for c in range(COLS - 2):
            if board[r][c] == board[r - 1][c + 1] == board[r - 2][c + 2] == player:
                return True
    return False

def is_board_full(board):
    return all(board[0][c] != '-' for c in range(COLS))
```

### 2. Search Algorithms: Minimax & Alpha–Beta Pruning

```python
minimax_search_count = 0

def minimax(board, depth, is_maximizing, max_depth=7):
    global minimax_search_count
    minimax_search_count += 1

    if check_win(board, 'O'):
        return 100 - depth
    if check_win(board, 'X'):
        return depth - 100
    if is_board_full(board) or depth == max_depth:
        return 0

    if is_maximizing:
        best_val = float('-inf')
        for c in range(COLS):
            r = get_lowest_empty_row(board, c)
            if r != -1:
                board[r][c] = 'O'
                val = minimax(board, depth + 1, False, max_depth)
                board[r][c] = '-'
                best_val = max(best_val, val)
        return best_val
    else:
        best_val = float('inf')
        for c in range(COLS):
            r = get_lowest_empty_row(board, c)
            if r != -1:
                board[r][c] = 'X'
                val = minimax(board, depth + 1, True, max_depth)
                board[r][c] = '-'
                best_val = min(best_val, val)
        return best_val

def get_best_move_minimax(board, max_depth=7):
    best_val = float('-inf')
    best_col = None
    for c in [2, 1, 3, 0]:
        r = get_lowest_empty_row(board, c)
        if r != -1:
            board[r][c] = 'O'
            val = minimax(board, 1, False, max_depth)
            board[r][c] = '-'
            if val > best_val:
                best_val = val
                best_col = c
    return best_col

alphabeta_search_count = 0

def alphabeta(board, depth, alpha, beta, is_maximizing, max_depth=7):
    global alphabeta_search_count
    alphabeta_search_count += 1

    if check_win(board, 'O'):
        return 100 - depth
    if check_win(board, 'X'):
        return depth - 100
    if is_board_full(board) or depth == max_depth:
        return 0

    if is_maximizing:
        best_val = float('-inf')
        for c in [2, 1, 3, 0]:
            r = get_lowest_empty_row(board, c)
            if r != -1:
                board[r][c] = 'O'
                val = alphabeta(board, depth + 1, alpha, beta, False, max_depth)
                board[r][c] = '-'
                best_val = max(best_val, val)
                alpha = max(alpha, val)
                if beta <= alpha:
                    break
        return best_val
    else:
        best_val = float('inf')
        for c in [2, 1, 3, 0]:
            r = get_lowest_empty_row(board, c)
            if r != -1:
                board[r][c] = 'X'
                val = alphabeta(board, depth + 1, alpha, beta, True, max_depth)
                board[r][c] = '-'
                best_val = min(best_val, val)
                beta = min(beta, val)
                if beta <= alpha:
                    break
        return best_val

def get_best_move_alphabeta(board, max_depth=7):
    best_val = float('-inf')
    best_col = None
    alpha = float('-inf')
    beta = float('inf')
    for c in [2, 1, 3, 0]:
        r = get_lowest_empty_row(board, c)
        if r != -1:
            board[r][c] = 'O'
            val = alphabeta(board, 1, alpha, beta, False, max_depth)
            board[r][c] = '-'
            if val > best_val:
                best_val = val
                best_col = c
            alpha = max(alpha, val)
    return best_col
```

### 3. Simulation Driver

```python
def play_connect3(algorithm="alphabeta", preset_moves=None, max_depth=7):
    global minimax_search_count, alphabeta_search_count
    if algorithm == "minimax":
        print("---- Connect 3 Using MinMax Algorithm ----")
        minimax_search_count = 0
    else:
        print("---- Connect 3 Using Alpha Beta Pruning ----")
        alphabeta_search_count = 0

    print("User is X")
    print("AI is O")
    print("Here is the empty board:")
    board = create_board()
    display_board(board)

    input_iter = iter(preset_moves) if preset_moves else None

    def get_user_choice():
        if input_iter:
            choice = next(input_iter)
            print(f"User Choice of Drop : {choice}")
            return choice
        raw = input("User Choice of Drop (1-4): ").strip()
        return int(raw)

    while True:
        user_col = get_user_choice() - 1
        ur = get_lowest_empty_row(board, user_col)
        board[ur][user_col] = 'X'
        display_board(board)

        if check_win(board, 'X'):
            print("User Won!")
            break
        if is_board_full(board):
            print("Game Over: Draw!")
            break

        if algorithm == "minimax":
            ai_col = get_best_move_minimax(board, max_depth)
            searches = minimax_search_count
        else:
            ai_col = get_best_move_alphabeta(board, max_depth)
            searches = alphabeta_search_count

        print(f"AI Thinking ... AI Choice : {ai_col + 1}")
        air = get_lowest_empty_row(board, ai_col)
        board[air][ai_col] = 'O'
        display_board(board)

        if check_win(board, 'O'):
            print(f"AI Won.... Number of Searches done in total : {searches:,}")
            break
        if is_board_full(board):
            print(f"Game Over: Draw! Searches done in total : {searches:,}")
            break
```

### 4. Running Benchmark Comparison

```python
sample_user_drops = [2, 4, 2]

print("=== 1. Plain Minimax Execution ===")
play_connect3(algorithm="minimax", preset_moves=sample_user_drops, max_depth=7)

print("\n=== 2. Alpha-Beta Pruning Execution ===")
play_connect3(algorithm="alphabeta", preset_moves=sample_user_drops, max_depth=7)

mm_count = minimax_search_count
ab_count = alphabeta_search_count
pruned = mm_count - ab_count
reduction_pct = (pruned / mm_count) * 100 if mm_count > 0 else 0

print("\n" + "=" * 60)
print(f"Plain Minimax Total Node Evaluations:   {mm_count:,}")
print(f"Alpha-Beta Pruning Node Evaluations:    {ab_count:,}")
print(f"Search Space Reduction:                 {reduction_pct:.2f}%")
print("=" * 60)
```

## Experiment 8: Knowledge-Based Rule Agent for Wumpus World
### Objectives:
1. Construct a random or custom **Wumpus World** ($N \times N$ grid).
2. Implement a **Rule-Based Agent** tracking `Visited`, `PitFree`, and `WumpusCandidates`.
3. Plan shortest safe paths using BFS and fire arrow when necessary.
4. Execute and verify the 3 benchmark test cases from the lab manual.

### 1. Wumpus World Environment Class

```python
import random
from collections import deque

class WumpusWorld:
    def __init__(self, n=4, wumpus_pos=None, pits=None, gold_pos=None, p=0.2, seed=None):
        self.n = n
        if seed is not None:
            random.seed(seed)

        all_cells = [(x, y) for x in range(1, n + 1) for y in range(1, n + 1) if (x, y) != (1, 1)]

        self.wumpus_pos = wumpus_pos if wumpus_pos is not None else random.choice(all_cells)
        self.pits = set(pits) if pits is not None else {c for c in all_cells if random.random() < p}

        if gold_pos is not None:
            self.gold_pos = gold_pos
        else:
            valid_gold = [c for c in all_cells if c not in self.pits and c != self.wumpus_pos]
            self.gold_pos = random.choice(valid_gold) if valid_gold else (n, n)

        self.wumpus_alive = True

    def get_neighbors(self, pos):
        x, y = pos
        nbrs = []
        for dx, dy in [(1, 0), (0, 1), (-1, 0), (0, -1)]:
            nx, ny = x + dx, y + dy
            if 1 <= nx <= self.n and 1 <= ny <= self.n:
                nbrs.append((nx, ny))
        return nbrs

    def sense_percepts(self, pos, just_screamed=False):
        nbrs = self.get_neighbors(pos)
        stench = (self.wumpus_pos in nbrs)
        breeze = any(p in nbrs for p in self.pits)
        glitter = (pos == self.gold_pos)
        percepts = []
        if stench: percepts.append("Stench")
        if breeze: percepts.append("Breeze")
        if glitter: percepts.append("Glitter")
        if just_screamed: percepts.append("Scream")
        return percepts

    def shoot_arrow(self, pos, direction):
        if not self.wumpus_alive:
            return False

        cx, cy = pos
        while True:
            if direction == 'RIGHT': cx += 1
            elif direction == 'LEFT': cx -= 1
            elif direction == 'UP': cy += 1
            elif direction == 'DOWN': cy -= 1

            if 1 <= cx <= self.n and 1 <= cy <= self.n:
                if (cx, cy) == self.wumpus_pos:
                    self.wumpus_alive = False
                    return True
            else:
                break
        return False

    def print_true_map(self):
        print(f"| y \\ x | " + " | ".join(str(x) for x in range(1, self.n + 1)) + " |")
        print("|:-----:|" + ":-:|" * self.n)
        for y in range(self.n, 0, -1):
            row = [f"| {y} "]
            for x in range(1, self.n + 1):
                cell = (x, y)
                symbols = []
                if cell == (1, 1): symbols.append("A")
                if cell == self.wumpus_pos and cell in self.pits: symbols.append("W+P")
                elif cell == self.wumpus_pos: symbols.append("W")
                elif cell in self.pits: symbols.append("P")
                if cell == self.gold_pos: symbols.append("G")
                if not symbols: symbols.append(".")
                row.append(f"| {'+'.join(symbols)} ")
            row.append("|")
            print("".join(row))
```

### 2. Rule-Based Reasoning Agent Class

```python
class RuleBasedAgent:
    def __init__(self, n=4):
        self.n = n
        self.pos = (1, 1)
        self.visited = set()
        self.pit_free = {(1, 1)}
        self.all_cells = {(x, y) for x in range(1, n + 1) for y in range(1, n + 1)}
        self.wumpus_candidates = self.all_cells - {(1, 1)}
        self.wumpus_dead = False
        self.arrow_used = False
        self.actions_log = []
        self.trace = []

    def get_neighbors(self, pos):
        x, y = pos
        nbrs = []
        for dx, dy in [(1, 0), (0, 1), (-1, 0), (0, -1)]:
            nx, ny = x + dx, y + dy
            if 1 <= nx <= self.n and 1 <= ny <= self.n:
                nbrs.append((nx, ny))
        return nbrs

    def is_safe(self, cell):
        return (cell in self.pit_free) and (self.wumpus_dead or (cell not in self.wumpus_candidates))

    def update_beliefs(self, percepts):
        self.visited.add(self.pos)
        if "Breeze" not in percepts:
            for nbr in self.get_neighbors(self.pos):
                self.pit_free.add(nbr)

        nbrs = set(self.get_neighbors(self.pos))
        if "Stench" in percepts:
            if not self.wumpus_dead:
                self.wumpus_candidates &= nbrs
        else:
            self.wumpus_candidates -= nbrs
            self.wumpus_candidates.discard(self.pos)

    def find_path_to_target(self, target):
        queue = deque([(self.pos, [self.pos])])
        visited_bfs = {self.pos}
        while queue:
            curr, path = queue.popleft()
            if curr == target:
                return path
            for nbr in self.get_neighbors(curr):
                if (nbr in self.visited or self.is_safe(nbr)) and nbr not in visited_bfs:
                    visited_bfs.add(nbr)
                    queue.append((nbr, path + [nbr]))
        return None

    def run(self, world):
        step = 0
        just_screamed = False

        while True:
            step += 1
            percepts = world.sense_percepts(self.pos, just_screamed)
            just_screamed = False
            self.update_beliefs(percepts)
            safe_cells = sorted([c for c in self.all_cells if self.is_safe(c)])

            if "Glitter" in percepts:
                self.actions_log.append("GRAB")
                self.trace.append({
                    'step': step, 'action': 'GRAB', 'pos': self.pos,
                    'percepts': percepts, 'safe_cells': safe_cells
                })
                shortest_path = self.find_path_to_target((1, 1))
                if shortest_path: shortest_path.reverse()
                return "POSSIBLE", self.actions_log, shortest_path, self.trace

            unvisited_safe = [c for c in self.all_cells if self.is_safe(c) and c not in self.visited]

            if unvisited_safe:
                best_path = None
                for candidate in unvisited_safe:
                    path = self.find_path_to_target(candidate)
                    if path:
                        if best_path is None or len(path) < len(best_path):
                            best_path = path

                if best_path and len(best_path) > 1:
                    next_pos = best_path[1]
                    dx = next_pos[0] - self.pos[0]
                    dy = next_pos[1] - self.pos[1]
                    act = "RIGHT" if dx == 1 else ("LEFT" if dx == -1 else ("UP" if dy == 1 else "DOWN"))
                    self.actions_log.append(act)
                    self.trace.append({
                        'step': step, 'action': f"MOVE {act} -> {next_pos}", 'pos': self.pos,
                        'percepts': percepts, 'safe_cells': safe_cells
                    })
                    self.pos = next_pos
                    continue

            if not self.arrow_used and self.wumpus_candidates:
                self.arrow_used = True
                shot_direction = None
                for direction in ['RIGHT', 'UP', 'LEFT', 'DOWN']:
                    cx, cy = self.pos
                    if direction == 'RIGHT': target = (cx + 1, cy)
                    elif direction == 'LEFT': target = (cx - 1, cy)
                    elif direction == 'UP': target = (cx, cy + 1)
                    elif direction == 'DOWN': target = (cx, cy - 1)
                    if target in self.wumpus_candidates:
                        shot_direction = direction
                        break

                if shot_direction is None:
                    sorted_cands = sorted(list(self.wumpus_candidates), key=lambda c: (abs(c[0]-self.pos[0]) + abs(c[1]-self.pos[1])))
                    for cand in sorted_cands:
                        if cand[1] == self.pos[1]:
                            shot_direction = "RIGHT" if cand[0] > self.pos[0] else "LEFT"
                            break
                        elif cand[0] == self.pos[0]:
                            shot_direction = "UP" if cand[1] > self.pos[1] else "DOWN"
                            break

                if shot_direction:
                    hit = world.shoot_arrow(self.pos, shot_direction)
                    act_label = f"SHOOT_{shot_direction}"
                    if hit:
                        self.wumpus_dead = True
                        self.wumpus_candidates.clear()
                        just_screamed = True
                        act_label += " (Scream)"
                    else:
                        cx, cy = self.pos
                        while True:
                            if shot_direction == "RIGHT": cx += 1
                            elif shot_direction == "LEFT": cx -= 1
                            elif shot_direction == "UP": cy += 1
                            elif shot_direction == "DOWN": cy -= 1
                            if 1 <= cx <= self.n and 1 <= cy <= self.n:
                                self.wumpus_candidates.discard((cx, cy))
                            else:
                                break

                    self.actions_log.append(act_label)
                    self.trace.append({
                        'step': step, 'action': act_label, 'pos': self.pos,
                        'percepts': percepts, 'safe_cells': safe_cells
                    })
                    continue

            self.trace.append({
                'step': step, 'action': 'TERMINATE (No safe moves)', 'pos': self.pos,
                'percepts': percepts, 'safe_cells': safe_cells
            })
            return "NOT POSSIBLE", self.actions_log, None, self.trace
```

## Experiment 9: Propositional Logic Engine & Model Checking (Wumpus Reasoning)
### Objectives:
1. Represent propositional sentences as **nested tuples**.
2. Implement **Sentence Evaluation** (`pl_true`) and **Truth-Table Printing**.
3. Implement **Entailment** (`tt_entails`), **Validity** (`is_valid`), and **Satisfiability** (`is_satisfiable`).
4. Apply propositional logic model checking to safe navigation in the **Wumpus World**.

### 1. Propositional Sentence Evaluation and Truth Table Formatter

```python
import itertools
from collections import deque

def pl_true(sentence, model):
    if isinstance(sentence, bool):
        return sentence
    if isinstance(sentence, str):
        if sentence in model:
            return model[sentence]
        raise KeyError(f"Symbol '{sentence}' not assigned in model: {model}")

    op = sentence[0]
    args = sentence[1:]

    if op == 'not':
        return not pl_true(args[0], model)
    elif op == 'and':
        return all(pl_true(a, model) for a in args)
    elif op == 'or':
        return any(pl_true(a, model) for a in args)
    elif op == 'implies':
        return (not pl_true(args[0], model)) or pl_true(args[1], model)
    elif op == 'iff':
        return pl_true(args[0], model) == pl_true(args[1], model)
    else:
        raise ValueError(f"Unknown operator: {op}")

def get_symbols(sentence):
    if isinstance(sentence, str):
        return {sentence}
    if isinstance(sentence, (list, tuple)):
        symbols = set()
        for elem in sentence[1:]:
            symbols |= get_symbols(elem)
        return symbols
    return set()

def sentence_to_str(s):
    if isinstance(s, str):
        return s
    op = s[0]
    if op == 'not':
        return f"~{sentence_to_str(s[1])}"
    elif op == 'and':
        return "(" + " & ".join(sentence_to_str(a) for a in s[1:]) + ")"
    elif op == 'or':
        return "(" + " | ".join(sentence_to_str(a) for a in s[1:]) + ")"
    elif op == 'implies':
        return f"({sentence_to_str(s[1])} => {sentence_to_str(s[2])})"
    elif op == 'iff':
        return f"({sentence_to_str(s[1])} <=> {sentence_to_str(s[2])})"
    return str(s)

def print_truth_table(sentence):
    symbols = sorted(list(get_symbols(sentence)))
    n = len(symbols)
    expr_str = sentence_to_str(sentence)

    header = " | ".join(f"{s:^5}" for s in symbols) + f" | {expr_str:^25}"
    separator = "-" * len(header)

    print(f"\nTruth Table for: {expr_str}")
    print(separator)
    print(header)
    print(separator)

    for values in itertools.product([True, False], repeat=n):
        model = dict(zip(symbols, values))
        res = pl_true(sentence, model)
        row_str = " | ".join(f"{'T' if v else 'F':^5}" for v in values) + f" | {'T' if res else 'F':^25}"
        print(row_str)
    print(separator + "\n")
```

### 2. Entailment, Validity, and Satisfiability Checkers

```python
def tt_entails(kb, alpha):
    if isinstance(kb, (list, set)):
        kb_sentence = ('and', *kb) if len(kb) > 1 else (list(kb)[0] if kb else True)
    else:
        kb_sentence = kb

    symbols = sorted(list(get_symbols(kb_sentence) | get_symbols(alpha)))

    for values in itertools.product([True, False], repeat=len(symbols)):
        model = dict(zip(symbols, values))
        if pl_true(kb_sentence, model):
            if not pl_true(alpha, model):
                return False
    return True

def is_valid(sentence):
    symbols = sorted(list(get_symbols(sentence)))
    for values in itertools.product([True, False], repeat=len(symbols)):
        model = dict(zip(symbols, values))
        if not pl_true(sentence, model):
            return False
    return True

def is_satisfiable(sentence):
    symbols = sorted(list(get_symbols(sentence)))
    for values in itertools.product([True, False], repeat=len(symbols)):
        model = dict(zip(symbols, values))
        if pl_true(sentence, model):
            return True
    return False
```

### 3. Running Lab Test Cases: Tautology, Entailment & Contradictions

### 4. Propositional Logic in Wumpus World Navigation

```python
class LogicWumpusAgent:
    def __init__(self, n=4):
        self.n = n
        self.pos = (1, 1)
        self.visited = set()
        self.proven_safe = {(1, 1)}
        self.pit_clauses = {}
        self.wumpus_clauses = {}

    def get_neighbors(self, pos):
        x, y = pos
        nbrs = []
        for dx, dy in [(0, 1), (1, 0), (0, -1), (-1, 0)]:
            nx, ny = x + dx, y + dy
            if 1 <= nx <= self.n and 1 <= ny <= self.n:
                nbrs.append((nx, ny))
        return nbrs

    def record_percepts(self, pos, percepts):
        x, y = pos
        self.visited.add(pos)
        nbrs = self.get_neighbors(pos)

        breeze_sym = f"B_{x}_{y}"
        pit_nbrs = [f"P_{nx}_{ny}" for nx, ny in nbrs]
        b_equiv = ('iff', breeze_sym, ('or', *pit_nbrs))
        b_fact = breeze_sym if "Breeze" in percepts else ('not', breeze_sym)
        self.pit_clauses[pos] = [b_equiv, b_fact]

        stench_sym = f"S_{x}_{y}"
        w_nbrs = [f"W_{nx}_{ny}" for nx, ny in nbrs]
        s_equiv = ('iff', stench_sym, ('or', *w_nbrs))
        s_fact = stench_sym if "Stench" in percepts else ('not', stench_sym)
        self.wumpus_clauses[pos] = [s_equiv, s_fact]

    def is_cell_provably_safe(self, target):
        if target in self.proven_safe:
            return True
        tx, ty = target
        p_target = f"P_{tx}_{ty}"
        w_target = f"W_{tx}_{ty}"

        relevant_pit_kb = []
        relevant_wumpus_kb = []
        for nbr in self.get_neighbors(target):
            if nbr in self.pit_clauses:
                relevant_pit_kb.extend(self.pit_clauses[nbr])
            if nbr in self.wumpus_clauses:
                relevant_wumpus_kb.extend(self.wumpus_clauses[nbr])

        if not relevant_pit_kb or not relevant_wumpus_kb:
            return False

        pit_free = tt_entails(relevant_pit_kb, ('not', p_target))
        wumpus_free = tt_entails(relevant_wumpus_kb, ('not', w_target))

        if pit_free and wumpus_free:
            self.proven_safe.add(target)
            return True
        return False

    def find_safe_path_to_gold(self, gold_pos, wumpus_pos, pits, max_steps=25):
        current = (1, 1)
        path_taken = [current]
        step = 0

        while current != gold_pos and step < max_steps:
            step += 1
            nbrs = self.get_neighbors(current)
            stench = (wumpus_pos in nbrs)
            breeze = any(p in nbrs for p in pits)
            percepts = []
            if stench: percepts.append("Stench")
            if breeze: percepts.append("Breeze")

            self.record_percepts(current, percepts)

            for nbr in nbrs:
                self.is_cell_provably_safe(nbr)

            unvisited_safe = [c for c in self.proven_safe if c not in self.visited]
            if not unvisited_safe:
                break

            queue = deque([(current, [current])])
            seen = {current}
            next_step = None

            while queue:
                node, path = queue.popleft()
                if node in unvisited_safe:
                    if len(path) > 1:
                        next_step = path[1]
                    break
                for n in self.get_neighbors(node):
                    if n in self.proven_safe and n not in seen:
                        seen.add(n)
                        queue.append((n, path + [n]))

            if next_step:
                current = next_step
                path_taken.append(current)
            else:
                break

        return path_taken

# Execute on Test 1 Map
agent = LogicWumpusAgent(n=4)
gold = (2, 3)
wumpus = (1, 3)
pits = {(3, 1), (3, 3), (4, 4)}

safe_path = agent.find_safe_path_to_gold(gold, wumpus, pits)
print("Discovered Path via Propositional Logic:")
print(" -> ".join(f"({x},{y})" for x, y in safe_path))
```
